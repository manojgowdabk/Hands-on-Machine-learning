```python
X = [1, 2, 3, 4, 5]  # Independent variable
y = [2, 4, 5, 4, 5]

mean_X = sum(X)/len(X)
mean_Y = sum(y)/len(y)

numerator = sum((X[i]-mean_X) * (y[i]-mean_Y) for i in range(len(X)))
denominator = sum((X[i] - mean_X) ** 2 for i in range(len(X)))

m = numerator / denominator
b = mean_Y -m * mean_X

print(f"slope (m): {m}")
print(f"Intercept (b): {b}")
```

```python
def predict(x):
    return m * x + b

predictions = [predict(x) for x in X]
print(f"prediction : {predictions}")
```

```python
import matplotlib.pyplot as plt
plt.scatter(X, y, color='blue', label='Data Points')
plt.plot(X, predictions, color = "red", label ="regression Line")
plt.xlabel('X')
plt.ylabel('y')
plt.title("linear regression")
plt.legend()
plt.show()
```

```python
# prompt: I wants to see the sgd in suitable visual

# Assume X_b and y are already defined from previous context

# Visualize the SGD learning process (conceptual)
# This requires the actual data points and the evolving model parameters (theta)
# Since the loop modifies 'theta' iteratively, we can plot the error or
# the model's prediction against the true values at each step.

# For simplicity, let's visualize the loss over epochs.
# We'll need to store the loss at each step or epoch.

loss_history = []
theta_history = [] # To store theta at each epoch

# Re-run the SGD loop to collect loss and theta history
theta = np.random.randn(2, 1)  # Reset theta for visualization
m = 10

for epoch in range(n_epochs):
    epoch_loss = 0
    for i in range(m):
        random_index = np.random.randint(m)
        xi = X_b[random_index : random_index + 1]
        yi = y[random_index : random_index + 1]
        predictions = xi.dot(theta)
        error = predictions - yi
        gradients = 2 * xi.T.dot(error)
        eta = learning_schedule(epoch * m + i)
        theta = theta - eta * gradients

        # Calculate and accumulate loss for the epoch
        epoch_loss += np.mean(error**2)

    loss_history.append(epoch_loss / m)  # Average loss per sample for the epoch
    theta_history.append(theta.copy())


# Plot the loss history
plt.figure(figsize=(10, 6))
plt.plot(range(n_epochs), loss_history)
plt.xlabel("Epoch")
plt.ylabel("Average Loss (Mean Squared Error)")
plt.title("SGD Regression Loss Over Epochs")
plt.grid(True)
plt.show()


# To visualize the model fitting the data over time, we need the data points (X_b, y)
# and the theta values from theta_history.
# This is more complex as it would involve plotting the regression line
# for different theta values during the training.

# Example: Plot the initial, middle, and final regression lines
plt.figure(figsize=(10, 6))
plt.scatter(X[:, 0], y, alpha=0.5, label="Data Points") # Assuming X is the feature

# Plot the initial regression line
theta_initial = theta_history[0]
plt.plot(
    X[:, 0],
    X_b.dot(theta_initial),
    color="red",
    linestyle="--",
    label=f"Epoch 0 (theta={theta_initial.flatten()})",
)

# Plot a middle regression line (e.g., at n_epochs/2)
if n_epochs > 1:
    middle_epoch = n_epochs // 2
    theta_middle = theta_history[middle_epoch]
    plt.plot(
        X[:, 0],
        X_b.dot(theta_middle),
        color="green",
        linestyle="-.",
        label=f"Epoch {middle_epoch} (theta={theta_middle.flatten()})",
    )

# Plot the final regression line
theta_final = theta_history[-1]
plt.plot(
    X[:, 0],
    X_b.dot(theta_final),
    color="purple",
    linestyle="-",
    label=f"Epoch {n_epochs-1} (theta={theta_final.flatten()})",
)


plt.xlabel("X (Feature)")
plt.ylabel("y (Target)")
plt.title("SGD Regression: Model Evolution Over Epochs")
plt.legend()
plt.grid(True)
plt.show()

```
